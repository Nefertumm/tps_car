```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
n_proc = np.array([2, 4, 8, 16])
bw_real_mbs = np.array([48154.9, 92352.2, 163805.0, 42897.9])
```

```python
b_base = bw_real_mbs[0]
bw_ideal_mbs = (n_proc / 2.0) * b_base
```

```python
plt.figure(figsize=(10, 6))
plt.plot(n_proc, bw_real_mbs, marker='o', linestyle='-', color='red', label='Ancho de Banda Real (Medido)')
plt.plot(n_proc, bw_ideal_mbs, marker='s', linestyle='--', color='blue', label=r'Ancho de Banda Teórico $((n/2) \cdot b)$')

plt.title('Ancho de Banda de Disección vs. Número de Procesadores')
plt.xlabel('Número de Procesadores (n)')
plt.ylabel('Ancho de Banda Agregado (MB/s)')
plt.xticks(n_proc)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.tight_layout()
```
