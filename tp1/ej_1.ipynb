```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```

```python
sizes: list[int] = [1, 1024, 16384, 1048576, 16777216, 134217728, 268435456, 536870912, 1073741824]
times: list[float] = [2.63333e-05, 3.60072e-05, 6.35207e-05, 0.00107675, 0.0145206, 0.11786, 0.23048, 0.460579, 0.921596]
```

```python
slope, intercept, r_value, p_value, std_err = linregress(sizes, times)
```

```python
latency = intercept
bandwidth = 1 / slope
```

```python
latency_us = latency * 1e6
bandwidth_Mbps = bandwidth * 8 / 1e6
```

```python
print(f'Latency: {latency_us:.2f} us, Bandwidth: {bandwidth_Mbps:.2f} Mb/s')
```

```python
plt.scatter(sizes, times, color='red', label='Datos medidos del cluster (avg)')
plt.plot(sizes, intercept + slope * np.array(sizes), color='blue', label='Ajuste lineal')
plt.xlabel('Tamaño de los datos')
plt.ylabel('Tiempo')
plt.legend()
plt.grid(True)
plt.show()
```
