```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```

```python
sizes = [1, 1024, 16384, 1048576, 16777216, 134217728, 268435456, 536870912, 1073741824]
times = [3.02775e-07, 1.11855e-06, 1.63933e-06, 2.4637e-05, 0.000331233, 0.00590523, 0.0131532, 0.0271056, 0.0553271]
```

```python
slope, intercept, r_value, p_value, std_err = linregress(sizes, times)
```

```python
latency = intercept
bandwidth = 1 / slope
```

```python
latency_us = latency * 1e6
bandwidth_Mbps = bandwidth * 8 / 1e6
```

```python
print(f'Latency: {latency_us:.2f} us, Bandwidth: {bandwidth_Mbps:.2f} Mb/s')
```

```python
plt.scatter(sizes, times, color='red', label='Datos medidos del cluster (avg)')
plt.plot(sizes, intercept + slope * np.array(sizes), color='blue', label='Ajuste lineal')
plt.xlabel('Tamaño de los datos')
plt.ylabel('Tiempo')
plt.legend()
plt.grid(True)
plt.show()
```
