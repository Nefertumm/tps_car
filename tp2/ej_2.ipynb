```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
procesadores = np.array([1, 2, 4, 8, 16, 32])
```

```python
tiempos_N512 = np.array([0.135918, 0.104663, 0.0583718, 0.0488224, 0.0369823, 0.0312781])
tiempos_N1024 = np.array([0.84357, 0.688212, 0.333551, 0.191663, 0.149252, 0.123374])
tiempos_N2048 = np.array([7.25367, 5.02159, 2.40836, 1.43873, 0.975539, 0.798974])
```

```python
def calcular_metricas(tiempos, procs):
    T1 = tiempos[0]
    speedup = T1 / tiempos
    eficiencia = speedup / procs
    return speedup, eficiencia
```

```python
S_512, E_512 = calcular_metricas(tiempos_N512, procesadores)
S_1024, E_1024 = calcular_metricas(tiempos_N1024, procesadores)
S_2048, E_2048 = calcular_metricas(tiempos_N2048, procesadores)
```

```python
fig, ax = plt.subplots(figsize=(8, 5), dpi=300)

# Speedup
ax.plot(procesadores, procesadores, 'k--', alpha=0.6, label='Speedup Ideal (Lineal)')
ax.plot(procesadores, S_512, marker='^', linewidth=2, label='N=512')
ax.plot(procesadores, S_1024, marker='o', linewidth=2, label='N=1024')
ax.plot(procesadores, S_2048, marker='s', linewidth=2, label='N=2048')

ax.set_title('Speedup vs Número de Procesadores')
ax.set_xlabel('Procesadores (p)')
ax.set_ylabel('Speedup (S)')
ax.set_xticks(procesadores)
ax.grid(True, linestyle=':', alpha=0.7)
ax.legend()


plt.tight_layout()
plt.savefig('ej2_speedup.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5), dpi=300)

# Eficiencia
ax.plot(procesadores, np.ones_like(procesadores), 'k--', alpha=0.6, label='Eficiencia Ideal (1.0)')
ax.plot(procesadores, E_512, marker='^', linewidth=2, label='N=512')
ax.plot(procesadores, E_1024, marker='o', linewidth=2, label='N=1024')
ax.plot(procesadores, E_2048, marker='s', linewidth=2, label='N=2048')

ax.set_title('Eficiencia vs Número de Procesadores')
ax.set_xlabel('Procesadores (p)')
ax.set_ylabel('Eficiencia (E)')
ax.set_xticks(procesadores)
ax.set_ylim(0, 1.1)
ax.grid(True, linestyle=':', alpha=0.7)
ax.legend()

plt.tight_layout()
plt.savefig('ej2_eficiencia.png', dpi=300, bbox_inches='tight')
plt.show()
```
