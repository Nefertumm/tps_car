```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
procesadores = np.array([1, 2, 4, 8, 16])
```

```python
tiempos_N1024 = np.array([10.5, 5.3, 2.7, 1.4, 0.8])
tiempos_N2048 = np.array([85.2, 42.8, 21.5, 11.0, 5.8])
```

```python
def calcular_metricas(tiempos, procs):
    T1 = tiempos[0]
    speedup = T1 / tiempos
    eficiencia = speedup / procs
    return speedup, eficiencia
```

```python
S_1024, E_1024 = calcular_metricas(tiempos_N1024, procesadores)
S_2048, E_2048 = calcular_metricas(tiempos_N2048, procesadores)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Speedup
ax1.plot(procesadores, procesadores, 'k--', alpha=0.6, label='Speedup Ideal (Lineal)')
ax1.plot(procesadores, S_1024, marker='o', linewidth=2, label='N=1024')
ax1.plot(procesadores, S_2048, marker='s', linewidth=2, label='N=2048')

ax1.set_title('Speedup vs Número de Procesadores')
ax1.set_xlabel('Procesadores (p)')
ax1.set_ylabel('Speedup (S)')
ax1.set_xticks(procesadores)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend()

# Eficiencia
ax2.plot(procesadores, np.ones_like(procesadores), 'k--', alpha=0.6, label='Eficiencia Ideal (1.0)')
ax2.plot(procesadores, E_1024, marker='o', linewidth=2, label='N=1024')
ax2.plot(procesadores, E_2048, marker='s', linewidth=2, label='N=2048')

ax2.set_title('Eficiencia vs Número de Procesadores')
ax2.set_xlabel('Procesadores (p)')
ax2.set_ylabel('Eficiencia (E)')
ax2.set_xticks(procesadores)
ax2.set_ylim(0, 1.1)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.legend()

plt.tight_layout()
```
